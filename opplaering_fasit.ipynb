### Blokk 4 — For-løkke som summerer
🎯 **Oppgave:** Løkka summerer lista og regner snittet. Snittet er nå **74** — det er akkurat Σ-formelen fra i går, i kode. Legg til **80** i lista og kjør på nytt. Snittet skal bli **75**.

````python
kvm = [45, 68, 95, 52, 110, 80]

sum_kvm = 0
for i in range(6):
    sum_kvm = sum_kvm + kvm[i]     # legg til ett og ett tall

snitt = sum_kvm / len(kvm)
print("Sum:", sum_kvm, "  Snitt:", snitt)

# Legg 80 inn i lista over og kjør på nytt -> snittet skal bli 75
````

💡 **Hint:** Sett 80 inn i klammene: `[45, 68, 95, 52, 110, 80]`. Løkka går automatisk ett steg ekstra, og `len()` teller det nye elementet med.

### Blokk 5 — If / else inni en løkke
🎯 **Oppgave:** Løkka merker hver leilighet som "stor" eller "liten" ut fra grensen 70. 1) Endre grensen fra 70 til 60. 2) Legg til en `elif` slik at leiligheter mellom 60 og 90 blir "middels".

````python
kvm = [45, 68, 95, 52, 110]

for k in kvm:
    if k > 90:
        print(k, "stor")
    elif k > 60:
        print(k, "middels")
    else:
        print(k, "liten")
````

### Blokk 6 — While-løkke
🎯 **Oppgave:** Løkka trekker fra 100 helt til verdien er ≤ 0. Endre startverdien til **1000**. Hvor mange steg tar den da?

**Fasit:** 500 → 5 steg. 1000 → 10 steg.

⏹ *Hvis en while-løkke aldri stopper: trykk stopp-knappen ved siden av cellen.*

````python
verdi = 1000
steg = 0
while verdi > 0:
    verdi = verdi - 100
    steg = steg + 1
    print("Steg", steg, "-> verdi", verdi)   # 10 steg
````

💡 **Hint:** `while` kjører så lenge betingelsen (`verdi > 0`) er sann. 1000 / 100 = 10 steg. Blir betingelsen aldri usann, kjører løkka i det uendelige — derfor finnes stopp-knappen.

### Blokk 7 — Funksjoner
🎯 **Oppgave:** Fyll inn hullet slik at funksjonen `storst(liste)` returnerer det største tallet. `storst(kvm)` skal gi **110**.

````python
kvm = [45, 68, 95, 52, 110]

def storst(liste):
    beste = liste[0]
    for x in liste:
        if x > beste:
            beste = x
    return beste

print(storst(kvm))   # 110
````

💡 **Hint:** Når `x` er større enn `beste`, skal `beste` oppdateres til `x`. Fyll inn `x` i hullet.

### Blokk 10 — NumPy: prediksjon og bom
🎯 **Oppgave:** Dette er **bom-tabellen fra i går**, i kode. `pred = kvm * w` er modellens gjett, og `bom = pris - pred` er hvor mye den bommer. Endre `w` fra `0.04` til `0.056` og kjør på nytt. **Ser prediksjonene mer ut som de faktiske prisene?**

````python
import numpy as np

kvm  = np.array([45, 68, 95, 52, 110])
pris = np.array([3.2, 4.1, 4.8, 4.0, 5.1])

w = 0.056              # endret fra 0.04
pred = kvm * w         # modellens gjett (mill)
bom  = pris - pred     # hvor mye vi bommer

print("Gjett:", pred)
print("Bom:  ", bom)
# Prediksjonene ligger nærmere de faktiske prisene enn med w=0.04
````

# Bolk 5 — Gradient descent

Nå setter vi gradient descent fra i går på ekte data: én feature (`Gr Liv Area`) skal predikere prisen. Tre deler.

### 5a — Kjør denne
🎯 **Oppgave:** Kjør cellen og følg med på **MSE**-kolonnen. Hva skjer?

**Fasit:** MSE starter på ca. **3.9e10** og *eksploderer*: 3.9e10 → 9.0e19 → 2.3e29 → ... → **3.6e57** etter seks steg. Modellen sprenger.

````python
import numpy as np

X = df['Gr Liv Area'].values.astype(float)
y = df['SalePrice'].values.astype(float)

w = 0.0
eta = 0.01
for steg in range(6):
    pred = w * X
    mse  = np.mean((y - pred)**2)
    grad = np.mean(-2 * X * (y - pred))
    w = w - eta * grad
    print(steg, f"w={w:.3e}  MSE={mse:.3e}")
````

💡 **Hint:** Se på tallene i MSE-kolonnen — blir de større eller mindre for hvert steg? Går det riktig vei?

### 5b — Finn ut hvorfor
🎯 **Oppgave:** Hvorfor eksploderer den? Ikke endre løkka ennå — bare undersøk størrelsene. Kjør cellen.

**Fasit:** Gradienten i første steg er ca. **-6.0e8** (minus 600 millioner!). `X` går opp til **5642** og `y` opp til **755 000** — store tall. Derfor tar `w = w - eta*grad` et gigantisk hopp forbi målet og bommer verre for hvert steg.

````python
# X og y er allerede definert i 5a
print("grad i første steg:", np.mean(-2 * X * (y - 0.0 * X)))
print("X max:", X.max())
print("y max:", y.max())
````

💡 **Hint:** `grad` blir enormt fordi `X` og `y` er store tall. `w = w - eta * grad` tar da et gigantisk hopp langt forbi målet — og bommer verre for hvert steg. Det er derfor det spretter fram og tilbake med voksende MSE.

### 5c — Få den til å virke
🎯 **Oppgave:** Få løkka til å konvergere. Fyll inn skaleringen og kjør. Koden er *ellers identisk* med 5a.

**Fasit:** MSE starter på ca. **3.91** og **synker** for hvert steg: 3.91 → 3.56 → 3.24 → 2.96 → 2.70 → 2.47. Grepet heter **normalisering**.

````python
X = df['Gr Liv Area'].values.astype(float) / 1000       # nå ~1.5
y = df['SalePrice'].values.astype(float) / 100000       # nå ~1.8

w = 0.0
eta = 0.01
for steg in range(6):
    pred = w * X
    mse  = np.mean((y - pred)**2)
    grad = np.mean(-2 * X * (y - pred))
    w = w - eta * grad
    print(steg, f"w={w:.3e}  MSE={mse:.3e}")   # MSE synker: 3.91 -> 2.47
````

💡 **Hint:** Del `X` på 1000 og `y` på 100000 før løkka. Da blir tallene små (~1.5 og ~1.8 i stedet for ~1500 og ~180000), og hoppene blir passe store. Dette grepet heter **normalisering**.
