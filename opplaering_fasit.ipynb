### Blokk 10 — NumPy: prediksjon og bom
🎯 **Oppgave:** Dette er **bom-tabellen fra i går**, i kode. `pred = kvm * w` er modellens gjett, og `bom = pris - pred` er hvor mye den bommer. Endre `w` fra `0.04` til `0.056` og kjør på nytt. **Ser prediksjonene mer ut som de faktiske prisene?**

**Fasit:** Med `w = 0.04` er alle bommene positive — modellen gjetter for lavt på hvert eneste hus.

Med `w = 0.056` ligger prediksjonene rundt de faktiske prisene: noen litt over, noen litt under. Snittet av bom² faller fra **1.81** til **0.63**.

Det er sånn en god modell ser ut. Den bommer i begge retninger i stedet for systematisk én vei.

````python
import numpy as np

kvm  = np.array([45, 68, 95, 52, 110])
pris = np.array([3.2, 4.1, 4.8, 4.0, 5.1])

w = 0.056           # endret fra 0.04
pred = kvm * w         # modellens gjett (mill)
bom  = pris - pred     # hvor mye vi bommer

print("Gjett:", pred)
print("Bom:  ", bom)
# Prediksjonene ligger nærmere de faktiske prisene enn med w=0.04
````

💡 **Hint:** Sammenlign `Gjett` med `pris` linje for linje. Med større `w` gjetter modellen høyere på alle hus. Ligger gjettene nærmere de faktiske prisene nå?

# Bolk 5 — Gradient descent

Nå setter vi gradient descent fra i går på ekte data: én feature (`Gr Liv Area`) skal predikere prisen. Tre deler.

### 5a — Kjør denne
🎯 **Oppgave:** Kjør cellen og følg med på **MSE**-kolonnen. Hva skjer?

**Fasit:** MSE starter på ca. **3.9e10** og *eksploderer*: 3.9e10 → 9.0e19 → 2.3e29 → ... → **3.6e57** etter seks steg. Modellen sprenger.

````python
import numpy as np

X = df['Gr Liv Area'].values.astype(float)
y = df['SalePrice'].values.astype(float)

w = 0.0
eta = 0.01
for steg in range(6):
    pred = w * X
    mse  = np.mean((y - pred)**2)
    grad = np.mean(-2 * X * (y - pred))
    w = w - eta * grad
    print(steg, f"w={w:.3e}  MSE={mse:.3e}")
````

💡 **Hint:** Se på tallene i MSE-kolonnen — blir de større eller mindre for hvert steg? Går det riktig vei?

### 5b — Finn ut hvorfor
🎯 **Oppgave:** Hvorfor eksploderer den? Ikke endre løkka ennå — bare undersøk størrelsene. Kjør cellen.

**Fasit:** Gradienten i første steg er ca. **-6.0e8** (minus 600 millioner!). `X` går opp til **5642** og `y` opp til **755 000** — store tall. Derfor tar `w = w - eta*grad` et gigantisk hopp forbi målet og bommer verre for hvert steg.

````python
# X og y er allerede definert i 5a
print("grad i første steg:", np.mean(-2 * X * (y - 0.0 * X)))
print("X max:", X.max())
print("y max:", y.max())
````

💡 **Hint:** `grad` blir enormt fordi `X` og `y` er store tall. `w = w - eta * grad` tar da et gigantisk hopp langt forbi målet — og bommer verre for hvert steg. Det er derfor det spretter fram og tilbake med voksende MSE.

### 🎯 Se hva som faktisk skjer

Kjør cellen. Den tegner **bollen** — altså MSE for alle mulige verdier av `w` — og markerer hvor de fire første stegene lander.

Til venstre: passe skrittlengde. Punktene ruller pent nedover mot bunnen.

Til høyre: for lang skrittlengde. Steg 1 hopper over bunnen og lander på den andre siden. Steg 2 hopper tilbake, men enda lenger. Punktene sprer seg utover i stedet for å nærme seg.

**Merk:** i 5a var `eta = 0.01`, og da lander steg 1 på seks millioner — så langt utenfor at det ikke kan tegnes. Figuren her bruker mindre skrittlengder for å vise det samme fenomenet på en skala vi ser.

````python
import numpy as np
import matplotlib.pyplot as plt

X = df['Gr Liv Area'].values.astype(float)
y = df['SalePrice'].values.astype(float)

def mse(w):
    return np.mean((y - w * X) ** 2)

def kjor(eta, n=4):
    w, sti = 0.0, [0.0]
    for _ in range(n):
        w = w - eta * np.mean(-2 * X * (y - w * X))
        sti.append(w)
    return sti

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, eta, farge, tittel, xlim in [
    (axes[0], 1e-7, "tab:blue", "Passe skritt: ruller ned", (-60, 320)),
    (axes[1], 6e-7, "tab:red",  "For langt skritt: spretter ut", (-2200, 2200)),
]:
    g = np.linspace(xlim[0], xlim[1], 400)
    ax.plot(g, [mse(v) for v in g], color="lightgray", lw=3)

    sti = kjor(eta)
    ys = [mse(v) for v in sti]
    ax.plot(sti, ys, "o-", color=farge, ms=10, lw=2)
    for i, (w, v) in enumerate(zip(sti, ys)):
        ax.annotate(str(i), (w, v), textcoords="offset points",
                    xytext=(0, 12), ha="center", color=farge, fontweight="bold")

    ax.axvline(119.65, color="black", ls=":", lw=1.5)
    ax.set_title(tittel, color=farge, fontweight="bold")
    ax.set_xlabel("w")
    ax.set_ylabel("MSE")
    ax.set_xlim(xlim)

plt.tight_layout()
plt.show()
````

### 5c — Få den til å virke
🎯 **Oppgave:** Få løkka til å konvergere. Fyll inn skaleringen og kjør. Koden er *ellers identisk* med 5a.

**Fasit:** MSE starter på ca. **3.91** og **synker** for hvert steg: 3.91 → 3.56 → 3.24 → 2.96 → 2.70 → 2.47. Grepet heter **normalisering**.

````python
X = df['Gr Liv Area'].values.astype(float) / 1000       # nå ~1.5
y = df['SalePrice'].values.astype(float) / 100000       # nå ~1.8

w = 0.0
eta = 0.01
for steg in range(6):
    pred = w * X
    mse  = np.mean((y - pred)**2)
    grad = np.mean(-2 * X * (y - pred))
    w = w - eta * grad
    print(steg, f"w={w:.3e}  MSE={mse:.3e}")   # MSE synker: 3.91 -> 2.47
````

💡 **Hint:** Et hus på 175 cm og et hus på 1.75 m er like høyt — tallet ble mindre, men høyden er den samme.
